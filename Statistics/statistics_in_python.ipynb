<p style="font-size:20px">As far as statistics is concerned, NumPy provides a variety of functions for calculating descriptive statistical measures such as mean, median, variance, and standard deviation. These functions can be applied to NumPy arrays to quickly and efficiently compute statistical summaries of the data.</p>
<p>Let us have a quick glance at these statistical measures:</p>
<ul style="font-size: 20px">
<li><strong>Mean:</strong> The mean is the average value of a dataset, calculated by summing all the values and dividing by the total number of values. Example: Suppose we have a dataset of exam scores: [85, 90, 78, 92, 88]. The mean score would be (85 + 90 + 78 + 92 + 88) / 5 = 86.6.</li><br/>
<li><strong>Median:</strong> The median is the middle value of a dataset when the values are arranged in order. If there is an even number of values, the median is the average of the two middle values.<br/> Now, median has two cases, if the number of data points are odd. Example: For the dataset [85, 90, 78, 92, 88], we have to first arrange in order [78, 85, 88, 90, 92], then apply the formula median = n+1/2 = 5+1/2 = 3rd value = 88.<br/> If the number of data points are even. Example: For the dataset [85, 90, 78, 92], we have to first arrange in order [78, 85, 90, 92], then apply the formula median = (n/2)th value + (n/2 +1)th value / 2 = (4/2)th value + (4/2 +1)th value / 2 = (2nd value + 3rd value) / 2 = (85 + 90) / 2 = 87.5.</li><br/>
<li><strong>Variance:</strong> Variance measures how much the values in a dataset vary from the mean. It is calculated by taking the average of the squared differences between each value and the mean. Example: For the dataset [85, 90, 78, 92, 88], the mean is 86.6. The formula for variance is <i>σ² = Σ (xi - μ)² / N</i>, where xi represents each value, μ is the mean, and N is the number of values. The variance would be [(85-86.6)² + (90-86.6)² + (78-86.6)² + (92-86.6)² + (88-86.6)²] / 5 = 23.04.</li><br/>
<li><strong>Standard Deviation:</strong> Standard deviation is the square root of the variance and provides a measure of how much the values in a dataset deviate from the mean on average. It is expressed in the same units as the original data, making it easier to interpret. Example: Using the variance calculated above (23.04), the standard deviation would be √23.04 ≈ 4.8.</li>
</ul>

```python
# 1. Mean in NumPy
arr1 = np.array([28,26,34,41,57,38])
print(np.mean(arr1))
```

```python
# 2. Median 
# NumPy internally detects odd/even data points on its own.
print(np.median(arr1)) # 34+38 / 2 --> 36
```

```python
# 3. Variance - np.var() function
print(np.var(arr1))
```

```python
# 4. Standard Deviation - np.std() function
print(np.std(arr1))
```

<p style="font-size:20px">Correlation and Covariance are two important statistical measures used to describe the relationship between two or more variables in a dataset.</p>
<ul style="font-size: 20px">
<li><strong>Correlation:</strong> Correlation measures the strength and direction of the linear relationship between two variables. It is expressed as a value between -1 and 1, where -1 indicates a perfect negative correlation, 0 indicates no correlation, and 1 indicates a perfect positive correlation. A positive correlation means that as one variable increases, the other variable tends to increase as well, while a negative correlation means that as one variable increases, the other variable tends to decrease. NumPy provides the numpy.corrcoef() function to calculate the correlation coefficient between two variables. The formula for correlation coefficient (Pearson's r) is <i>r = Σ (xi - x̄)(yi - ȳ) / √[Σ (xi - x̄)² * Σ (yi - ȳ)²]</i>, where xi and yi are the individual data points, and x̄ and ȳ are the means of the respective variables.</li><br/>
<li><strong>Covariance:</strong> Covariance measures how two variables change together. It indicates the direction of the linear relationship between the variables, but not the strength. A positive covariance means that as one variable increases, the other variable tends to increase as well, while a negative covariance means that as one variable increases, the other variable tends to decrease. However, unlike correlation, covariance is not standardized and can take on any value. NumPy provides the numpy.cov() function to calculate the covariance between two variables. The formula for covariance is <i>cov(X, Y) = Σ (xi - x̄)(yi - ȳ) / (N - 1)</i>, where xi and yi are the individual data points, x̄ and ȳ are the means of the respective variables, and N is the number of data points.</li>
</ul>


```python
np.corrcoef([1,6,3], [10,20,30])
```

```python
np.cov([1,6,3],[10,20,30])
```

<p style="font-size:20px">Discrete Probability Distributions:</p>
<p style="font-size:20px">Discrete probability distributions describe the probabilities of outcomes for discrete random variables, which can take on a finite or countably infinite number of values. Examples of discrete probability distributions include the binomial distribution, Poisson distribution, and geometric distribution.</p>
<ul style="font-size: 20px">
<li><strong>Binomial Distribution:</strong> The binomial distribution models the number of successes in a fixed number of independent Bernoulli trials, where each trial has two possible outcomes (success or failure). It is characterized by two parameters: the number of trials (n) and the probability of success (p) in each trial. The probability mass function (PMF) of the binomial distribution is given by <i>P(X = k) = C(n, k) * p^k * (1-p)^(n-k)</i>, where C(n, k) is the binomial coefficient representing the number of ways to choose k successes from n trials.</li><br/>
<li><strong>Poisson Distribution:</strong> The Poisson distribution models the number of events occurring within a fixed interval of time or space, given a known average rate of occurrence (λ). It is characterized by a single parameter, λ, which represents the average number of events in the interval. The probability mass function (PMF) of the Poisson distribution is given by <i>P(X = k) = (λ^k * e^(-λ)) / k!</i>, where k is the number of events, e is Euler's number (approximately 2.71828), and k! is the factorial of k.</li><br/>
</ul>

```python
# Binomial Distribution
import numpy as np
from scipy.stats import binom
import matplotlib.pyplot as plt

n = 3
p = 0.5

x = np.arange(0, n+1)
pmf = binom.pmf(x, n, p)

plt.stem(x, pmf, use_line_collection=True)
plt.xlabel("Number of tails (X)")
plt.ylabel("Probability")
plt.title("PMF of Number of Tails in 3 Coin Tosses")
plt.show()
```

```python
# Poisson distribution
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

# Parameter (lambda = average rate)
lam = 3

# Range of x values (0 to 15 events)
x = np.arange(0, 16)

# PMF and CDF
pmf = poisson.pmf(x, lam)
cdf = poisson.cdf(x, lam)

# ---- PMF Plot ----
plt.stem(x, pmf, use_line_collection=True)
plt.title("Poisson Distribution - PMF (λ = 3)")
plt.xlabel("Number of events (k)")
plt.ylabel("P(X = k)")
plt.show()

# ---- CDF Plot ----
plt.step(x, cdf, where='post')
plt.title("Poisson Distribution - CDF (λ = 3)")
plt.xlabel("Number of events (k)")
plt.ylabel("P(X ≤ k)")
plt.show()
```

<p style="font-size:20px">Continuous Probability Distributions:</p>
<p style="font-size:20px">Continuous probability distributions describe the probabilities of outcomes for continuous random variables, which can take on any value within a given range. Examples of continuous probability distributions include the normal distribution, exponential distribution, and uniform distribution.</p>
<ul style="font-size: 20px">
<li><strong>Normal Distribution:</strong> The normal distribution, also known as the Gaussian distribution 
, is a continuous probability distribution characterized by its bell-shaped curve. It is defined by two parameters: the mean (μ) and the standard deviation (σ). The probability density function (PDF) of the normal distribution is given by <i>f(x) = (1 / (σ * √(2π))) * e^(-0.5 * ((x - μ) / σ)^2)</i>, where e is Euler's number (approximately 2.71828) and π is Pi (approximately 3.14159). The normal distribution is widely used in statistics and data science due to its prevalence in natural phenomena and its mathematical properties.</li><br/>
<li><strong>Exponential Distribution:</strong> The exponential distribution models the time between events in a Poisson process, where events occur continuously and independently at a constant average rate (λ). It is characterized by a single parameter, λ, which represents the rate of occurrence. The probability density function (PDF) of the exponential distribution is given by <i>f(x) = λ * e^(-λx)</i> for x ≥ 0, where e is Euler's number (approximately 2.71828).</li><br/>
<li><strong>Uniform Distribution:</strong> The uniform distribution is a continuous probability distribution where all outcomes within a specified range are equally likely. It is characterized by two parameters: the minimum value (a) and the maximum value (b). The probability density function (PDF) of the uniform distribution is given by <i>f(x) = 1 / (b - a)</i> for a ≤ x ≤ b, and f(x) = 0 otherwise.</li>
<li>Cumulative Distribution Function (CDF): The cumulative distribution function (CDF) describes the probability that a random variable takes on a value less than or equal to a specific value. It is defined for both discrete and continuous probability distributions. The CDF is useful for calculating probabilities and percentiles, as well as for visualizing the distribution of data. It means what is the probability that x would be less than or equal to a particular value.</li>

<p style="font-size:20px">Now lets implement these distributions using SciPy:</p>

```python
from scipy import stats
import numpy as np
import matplotlib.pyplot as plt
# Example: Normal Distribution
mu = 0  # mean
sigma = 1  # standard deviation
x = np.linspace(-5, 5, 1000) # 1000 values between -5 and 5
pdf = stats.norm.pdf(x, mu, sigma) 
plt.plot(x, pdf)
plt.title('Normal Distribution (μ=0, σ=1)')
plt.xlabel('x')
plt.ylabel('Probability Density')
plt.show()
```

```python
# Exponential Distribution
from scipy.stats import expon

x = np.linspace(0, 10, 500)  # 500 values between 0 and 10 at regular intervals
pdf = expon.pdf(x, scale=1) # pdf - Probability DEnsity Function
cdf = expon.cdf(x, scale=1) # cdf - Cumulative Density Function 

plt.plot(x, pdf, label="PDF") 
plt.plot(x, cdf, label="CDF")
plt.title("Exponential Distribution")
plt.legend()
plt.show()

```

```python
# Uniform Distribution
from scipy.stats import uniform

x = np.linspace(0, 1, 500)
pdf = uniform.pdf(x, loc=0, scale=1)
cdf = uniform.cdf(x, loc=0, scale=1)

plt.plot(x, pdf, label="PDF")
plt.plot(x, cdf, label="CDF")
plt.title("Uniform Distribution")
plt.legend()
plt.show()
```

<h5>Standardization and Z-Scores</h5>
<p style="font-size:20px">Standardization is a statistical technique used to transform data to have a mean of 0 and a standard deviation of 1. This process is essential for comparing data points from different distributions or scales, as it allows for a common reference point.</p>
<p style="font-size:20px">Z-scores are a measure of how many standard deviations a data point is from the mean of the dataset. They are calculated using the formula:</p>
<i>Z = (X - μ) / σ</i>
<p style="font-size:20px">Where:<br/>
X = Individual data point<br/>
μ = Mean of the dataset<br/>
σ = Standard deviation of the dataset</p>
<p style="font-size:20px">By converting data points to z-scores, we can easily identify outliers, compare values across different datasets, and perform various statistical analyses that require standardized data.</p>

```python
import numpy as np
from scipy.stats import zscore

data = [10, 12, 15, 20, 22]

# Using scipy
z_scores = zscore(data)
print(z_scores)

# Manual
mean = np.mean(data)
std = np.std(data)
z_manual = [(x - mean)/std for x in data]
print(z_manual)
```

<h5>Skewness and Kurtosis</h5>
<p style="font-size:20px">Skewness and kurtosis are two important statistical measures used to describe the shape of a probability distribution.</p>
<ul style="font-size: 20px">
<li><strong>Skewness:</strong> Skewness measures the asymmetry of a probability distribution around its mean. A distribution is said to be positively skewed (right-skewed) if it has a longer tail on the right side, indicating that there are more extreme values on the higher end of the distribution. Conversely, a distribution is negatively skewed (left-skewed) if it has a longer tail on the left side, indicating more extreme values on the lower end. A skewness value of 0 indicates a perfectly symmetrical distribution.</li><br/>
<li><strong>Kurtosis:</strong> Kurtosis measures the "tailedness" of a probability distribution, indicating the presence of outliers. A distribution with high kurtosis has heavier tails and a sharper peak compared to a normal distribution, while a distribution with low kurtosis has lighter tails and a flatter peak. The kurtosis of a normal distribution is 3, and excess kurtosis is calculated by subtracting 3 from the kurtosis value. Positive excess kurtosis indicates a leptokurtic distribution (heavy tails), while negative excess kurtosis indicates a platykurtic distribution (light tails).</li>
</ul>

```python
# Skewness (Visualized)

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import skewnorm, norm

# ----- 1. Generate Data -----

# Right-Skewed (positive skew) → a > 0
right_skewed = skewnorm.rvs(a=6, loc=0, scale=1, size=5000)

# Left-Skewed (negative skew) → a < 0
left_skewed = skewnorm.rvs(a=-6, loc=0, scale=1, size=5000)

# Normal Distribution (no skew)
normal_data = norm.rvs(loc=0, scale=1, size=5000)

# ----- 2. Plot -----

plt.figure(figsize=(15, 4))

# Right Skewed
plt.subplot(1, 3, 1)
plt.hist(right_skewed, bins=40, density=True)
plt.title("Right-Skewed Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")

# Left Skewed
plt.subplot(1, 3, 2)
plt.hist(left_skewed, bins=40, density=True)
plt.title("Left-Skewed Distribution")
plt.xlabel("Values")

# Normal
plt.subplot(1, 3, 3)
plt.hist(normal_data, bins=40, density=True)
plt.title("Normal Distribution")
plt.xlabel("Values")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import t, norm, uniform, kurtosis

# Generate data
high_kurt = t.rvs(df=2, size=5000)           # heavy tails
normal = norm.rvs(loc=0, scale=1, size=5000) # normal kurtosis
low_kurt = uniform.rvs(loc=-2, scale=4, size=5000)  # light tails

# Plot
plt.figure(figsize=(15, 4))

# High Kurtosis
plt.subplot(1, 3, 1)
plt.hist(high_kurt, bins=50, density=True)
plt.title("High Kurtosis (Heavy Tails)\nK = {:.2f}".format(kurtosis(high_kurt)))

# Normal
plt.subplot(1, 3, 2)
plt.hist(normal, bins=50, density=True)
plt.title("Normal Distribution\nK = {:.2f}".format(kurtosis(normal)))

# Low Kurtosis
plt.subplot(1, 3, 3)
plt.hist(low_kurt, bins=50, density=True)
plt.title("Low Kurtosis (Flat)\nK = {:.2f}".format(kurtosis(low_kurt)))

plt.tight_layout()
plt.show()
```

<h5>Statistical Hypothesis Testing</h5>
<p style="font-size:20px">Statistical hypothesis testing is a method used to make inferences about a population based on a sample of data. It involves formulating a null hypothesis (H0) and an alternative hypothesis (H1), and then using statistical tests to determine whether to reject or fail to reject the null hypothesis.</p>
<ul style="font-size: 20px">
<li><strong>Null Hypothesis (H0):</strong> The null hypothesis is a statement that there is no effect or no difference between groups. It serves as the default assumption that researchers aim to test against.</li><br/>
<li><strong>Alternative Hypothesis (H1):</strong> The alternative hypothesis is a statement that contradicts the null hypothesis, suggesting that there is an effect or a difference between groups. Researchers seek to provide evidence in support of the alternative hypothesis.</li><br/>
<li><strong>Significance Level (α):</strong> The significance level is a threshold used to determine whether to reject the null hypothesis. It represents the probability of making a Type I error (rejecting a true null hypothesis). Common significance levels include 0.05, 0.01, and 0.10.</li><br/>
<li>Example: Suppose a researcher wants to test whether a new drug is more effective than a placebo in reducing blood pressure. The null hypothesis (H0) would state that there is no difference in blood pressure reduction between the drug and the placebo, while the alternative hypothesis (H1) would state that the drug is more effective than the placebo. The researcher would then collect data from a sample of patients, perform a statistical test (e.g., t-test), and compare the p-value to the significance level (α) to determine whether to reject or fail to reject the null hypothesis.</li><br/>
<li><strong>Common Statistical Tests:</strong> Some common statistical tests used in hypothesis testing include the t-test, chi-square test, ANOVA (analysis of variance), and regression analysis. Each test has its own assumptions and applications, depending on the type of data and research question.</li><br/>
<li><strong>P-Value:</strong> The p-value is a measure of the strength of evidence against the null hypothesis. It represents the probability of observing the data, or something more extreme, if the null hypothesis is true. A smaller p-value indicates stronger evidence against the null hypothesis. If the p-value is less than or equal to the significance level (α), the null hypothesis is rejected in favor of the alternative hypothesis. The general rule of thumb is:<br/>
- If p-value ≤ α: Reject H0 (evidence supports H1)<br/>
- If p-value > α: Fail to reject H0 (insufficient evidence to support H1)</li>
</ul>

<h5>Statistical Tests</h5>
<p style="font-size:20px">Statistical tests are used to analyze data and draw conclusions about populations based on sample data. They help determine whether observed differences or relationships in the data are statistically significant or likely due to random chance.</p>
<ul style="font-size: 20px">
<li><strong>T-Test:</strong> The t-test is used to compare the means of two groups to determine if there is a significant difference between them. It is commonly used in experiments where researchers want to assess the effect of a treatment or intervention. There are two main types of t-tests: independent t-test (for comparing two independent groups) and paired t-test (for comparing two related groups).</li><br/>
<li><strong>Chi-Square Test:</strong> The chi-square test is used to analyze categorical data and determine if there is a significant association between two or more categorical variables. It is commonly used in surveys and observational studies to assess relationships between variables.</li><br/>
<li><strong>ANOVA (Analysis of Variance):</strong> ANOVA is used to compare the means of three or more groups to determine if there is a significant difference among them. It is an extension of the t-test and is commonly used in experiments with multiple treatment groups.</li><br/>
</ul>

```python
# T-test (Two Sample)
# Case 1 - A teacher wants to check if Class A and Class B have significantly different average math scores.
# If p-value < 0.05 → There is a statistically significant difference between the average scores of Class A and Class B.
# If p-value ≥ 0.05 → There is no evidence to say the two classes have different mean scores.
import numpy as np
from scipy.stats import ttest_ind

# Scores of two different classes
class_A = np.array([78, 85, 90, 88, 76, 92])
class_B = np.array([72, 80, 75, 70, 68, 74])

t_stat, p_value = ttest_ind(class_A, class_B)

print("Independent T-Test")
print("T-statistic:", t_stat)
print("P-value:", p_value)
```

```python
# T test (Paired)
# Case 2 - We test whether a training program improved employee performance scores.
# (Measure before training and after training for the same people.)
# If p-value < 0.05 → The training program had a significant positive effect; employees performed better after training.
# If p-value ≥ 0.05 → There is no significant improvement after training.
from scipy.stats import ttest_rel
import numpy as np

before = np.array([60, 65, 70, 62, 68])
after  = np.array([72, 75, 78, 70, 74])

t_stat, p_value = ttest_rel(before, after)

print("Paired T-Test")
print("T-statistic:", t_stat)
print("P-value:", p_value)
```

```python
# One Way ANOVA
# Case - A nutritionist checks whether 3 different diets produce different weight loss results.
# If p-value < 0.05 → At least one diet leads to significantly different weight loss compared to others.
# If p-value ≥ 0.05 → All diets have similar performance; no significant difference.
from scipy.stats import f_oneway
import numpy as np

diet1 = np.array([2, 3, 2, 4, 3])
diet2 = np.array([5, 6, 7, 5, 6])
diet3 = np.array([1, 2, 1, 1, 2])

f_stat, p_value = f_oneway(diet1, diet2, diet3)

print("ANOVA Test")
print("F-statistic:", f_stat)
print("P-value:", p_value) # 0.00000278
```

```python
# Chi Squared Test
# A company wants to check whether Gender and Product Preference are related.
# 
# Gender	Likes Product	Doesn’t Like
# Male	        30	            20
# Female	    50	            10

# If p-value < 0.05 → Gender and Product Preference are statistically dependent → preference differs by gender.
# If p-value ≥ 0.05 → Gender and Product Preference are independent → gender does not influence preference.

import numpy as np
from scipy.stats import chi2_contingency

# Contingency Table
data = np.array([
    [30, 20],   # Male
    [50, 10]    # Female
])

chi2, p_value, dof, expected = chi2_contingency(data)

print("Chi-Squared Test")
print("Chi2:", chi2)
print("P-value:", p_value)
print("DOF:", dof)
print("Expected Table:\n", expected)
```
